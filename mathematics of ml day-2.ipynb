###  Determinant

```python
import numpy as np
```

```python
a = np.array([[2,3],[1,4]])
determinant = np.linalg.det(a)
print("determinant:",determinant)
```

### inverse of matrix

```python
inverse = np.linalg.inv(a)
print("inverse of matrix:" ,inverse)
```

### eigenvalue , eigenvector

```python
eigen_value , eigen_vector = np.linalg.eig(a),np.linalg.eigvals(a)
```

```python
print("eigen_value:",eigen_value)
```

```python
print("eigen_value:",eigen_vector)
```

4. SVD

```python
A = np.array([
    [3, 2],
    [1, 4]
])

U, S, Vt = np.linalg.svd(A)

print("U:")
print(U)

print("S:")
print(S)

print("Vt:")
print(Vt)
```
